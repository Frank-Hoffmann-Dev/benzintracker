```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATABASE_PATH = Path("/home/frankh/.local/share/benzintracker/benzintracker.db")

def get_connection(db_path: Path):
    return sqlite3.connect(db_path)

def get_all_data() -> pd.DataFrame:
    conn = get_connection(DATABASE_PATH)

    query = """
        SELECT * from stations s
        JOIN prices p ON p.station_id == s.id
    """

    df = pd.read_sql_query(query, conn, parse_dates=["recorded_at"])
    conn.close()

    return df


data = get_all_data()
```
